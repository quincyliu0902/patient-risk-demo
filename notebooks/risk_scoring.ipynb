```python
import sys
import pandas as pd
import numpy as np
import sklearn
import matplotlib.pyplot as plt
import seaborn as sns

print("Python executable:", sys.executable)
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
print("Scikit-learn version:", sklearn.__version__)

# Generate simple test data
df = pd.DataFrame({
    "patient_id": ["P001", "P002", "P003"],
    "ed_visits": [3, 0, 2],
    "total_cost": [12000, 2000, 8500]
})

display(df)

sns.scatterplot(data=df, x="ed_visits", y="total_cost")
plt.show()
```

```python
def calculate_risk(ed_visits, chronic_conditions):
    if ed_visits >= 2:
        return "High"
    elif ed_visits >= 1 or chronic_conditions >= 2:
        return "Medium"
    else:
        return "Low"
```

```python
# Add chronic condition information
df["chronic_conditions"] = [2, 1, 3]

# Apply the scoring function
df["risk_category"] = df.apply(
    lambda row: calculate_risk(
        row["ed_visits"],
        row["chronic_conditions"]
    ),
    axis=1
)

display(df)
```
